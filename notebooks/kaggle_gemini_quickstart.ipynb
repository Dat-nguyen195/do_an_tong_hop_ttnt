# SEMSYNTH Lite — Kaggle + Gemini
Add `GEMINI_API_KEY` in **Kaggle → Add-ons → Secrets**, enable Internet, then run the cell below. Replace `REPO_URL` with your GitHub URL.

```python
import os, subprocess, sys
from pathlib import Path
from kaggle_secrets import UserSecretsClient

REPO_URL = 'https://github.com/<YOUR_GITHUB_USERNAME>/<YOUR_REPO>.git'
WORKDIR = Path('/kaggle/working/semsynth-lite')

if not WORKDIR.exists():
    subprocess.run(['git', 'clone', REPO_URL, str(WORKDIR)], check=True)
os.chdir(WORKDIR)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'], check=True)

os.environ['GEMINI_API_KEY'] = UserSecretsClient().get_secret('GEMINI_API_KEY')
os.environ['LLM_PROVIDER'] = 'gemini'
os.environ['DEFAULT_MODEL'] = 'gemini-3.8-flash'

subprocess.run([sys.executable, '-m', 'pytest', '-q'], check=True)
subprocess.run([sys.executable, 'kaggle_run.py', '--offline-suite'], check=True)
subprocess.run([sys.executable, 'kaggle_run.py', '--provider', 'gemini', '--model', 'gemini-3.8-flash'], check=True)
```
